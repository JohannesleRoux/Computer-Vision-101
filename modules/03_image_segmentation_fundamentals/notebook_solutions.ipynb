```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
class TverskyLoss(nn.Module):
    """Tversky loss for binary segmentation."""
    def __init__(self, alpha=0.3, beta=0.7, smooth=1e-6):
        super().__init__()
        self.alpha, self.beta, self.smooth = alpha, beta, smooth

    def forward(self, logits, targets):
        probs   = torch.sigmoid(logits)
        targets = targets.float()
        tp = (probs * targets).sum(dim=(1,2,3))
        fp = (probs * (1 - targets)).sum(dim=(1,2,3))
        fn = ((1 - probs) * targets).sum(dim=(1,2,3))
        ti = (tp + self.smooth) / (tp + self.alpha * fp + self.beta * fn + self.smooth)
        return (1 - ti).mean()

# Verify
logits  = torch.randn(2, 1, 64, 64)
targets = (torch.rand(2, 1, 64, 64) > 0.5).float()
loss = TverskyLoss(alpha=0.3, beta=0.7)(logits, targets)
print(f'Tversky loss: {loss.item():.4f}')
```
