## 1. YOLOv1 Grid Cell Formulation

The image is divided into S×S cells. Each cell predicts B boxes and C class scores.

```python
S, B, C = 7, 2, 20
output_size = S * S * (B * 5 + C)
print(f'YOLOv1 output: {S}x{S}x{B*5+C} = {output_size} values')

# Visualise the grid
img = np.ones((448, 448, 3), dtype=np.uint8) * 200
fig, ax = plt.subplots(1, 1, figsize=(5, 5))
ax.imshow(img)
cell_size = 448 // S
for i in range(S + 1):
    ax.axhline(i * cell_size, color='gray', linewidth=0.5)
    ax.axvline(i * cell_size, color='gray', linewidth=0.5)
ax.set_title(f'YOLOv1: {S}x{S} grid on 448x448')
ax.axis('off')
plt.tight_layout(); plt.show()
```

## 2. Anchor Box K-Means

YOLOv2 introduced anchor boxes derived by k-means clustering on the training set.

```python
# Simulate GT box dimensions (normalised)
np.random.seed(42)
gt_wh = np.concatenate([
    np.random.normal([0.05, 0.15], 0.01, (200, 2)),  # pedestrians (narrow/tall)
    np.random.normal([0.20, 0.10], 0.02, (150, 2)),  # cars (wide/short)
    np.random.normal([0.08, 0.08], 0.01, (100, 2)),  # small objects
]).clip(0.01, 1.0)

anchors = anchor_kmeans(gt_wh, k=9)
print('Learned anchors (w, h):')
for i, (w, h) in enumerate(anchors):
    print(f'  {i+1}: w={w:.3f}, h={h:.3f}, ratio={h/w:.2f}')

# Visualize
fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(gt_wh[:,0], gt_wh[:,1], alpha=0.3, s=5, label='GT boxes')
ax.scatter(anchors[:,0], anchors[:,1], s=200, marker='*', color='red', zorder=5, label='Anchors')
ax.set_xlabel('Width'); ax.set_ylabel('Height')
ax.legend(); ax.set_title('Anchor K-Means Clustering')
plt.tight_layout(); plt.show()
```
