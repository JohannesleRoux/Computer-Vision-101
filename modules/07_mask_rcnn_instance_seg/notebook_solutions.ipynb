# Module 07 — Mask R-CNN (SOLUTIONS)

```python
import torch

def mask_iou(pred_mask: torch.Tensor, gt_mask: torch.Tensor) -> float:
    """Binary mask IoU."""
    pred = pred_mask.bool()
    gt   = gt_mask.bool()
    inter = (pred & gt).sum().float()
    union = (pred | gt).sum().float()
    return (inter / (union + 1e-6)).item()

# Test
pred = torch.zeros(10, 10, dtype=torch.bool)
pred[2:8, 2:8] = True
gt   = torch.zeros(10, 10, dtype=torch.bool)
gt[3:9, 3:9]   = True
print(f'Mask IoU: {mask_iou(pred, gt):.4f}')
print(f'Perfect overlap: {mask_iou(pred, pred):.4f}')
```
