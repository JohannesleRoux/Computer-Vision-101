# Module 08 — Panoptic Segmentation (SOLUTIONS)

```python
import numpy as np
import matplotlib.pyplot as plt
from panoptic_utils import merge_semantic_instance

semantic_map = np.zeros((128, 128), dtype=np.int64)
semantic_map[:64, :]    = 1  # sky
semantic_map[64:, :40]  = 2  # road
semantic_map[64:, 40:80]= 3  # person
semantic_map[64:, 80:]  = 4  # car

instance_map = np.zeros((128, 128), dtype=np.int64)
instance_map[64:, 40:80] = 1
instance_map[64:, 80:]   = 2  # two separate car instances

thing_ids = [3, 4]
panoptic_map, segments_info = merge_semantic_instance(semantic_map, instance_map, thing_ids)

print('Segments:')
for s in segments_info:
    print(f'  id={s["id"]}, cat={s["category_id"]}, is_thing={s["is_thing"]}')

# Visualize
colors = plt.cm.tab10(np.linspace(0, 1, len(segments_info)))
vis = np.zeros((*panoptic_map.shape, 3), dtype=np.uint8)
for i, seg in enumerate(segments_info):
    mask = panoptic_map == seg['id']
    vis[mask] = (np.array(colors[i][:3]) * 255).astype(np.uint8)

plt.figure(figsize=(6,6))
plt.imshow(vis)
plt.title('Panoptic Map')
plt.axis('off')
plt.tight_layout(); plt.show()
```
